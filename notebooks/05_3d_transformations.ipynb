# Notebook 05 - 3D Transformation


```python
import numpy as np

def rotation_z(angle_deg):

    theta = np.radians(angle_deg)

    return np.array([
        [np.cos(theta), -np.sin(theta), 0.0],
        [np.sin(theta), np.cos(theta), 0.0],
        [0.0,            0.0,           0.0]
    ])
```

```python
point = np.array([10.0, 0.0, 0.0])

R = rotation_z(90)

rotated = R @ point

print(rotated)
```

```python
translation = np.array([
    100.0,
    200.0,
    300.0
])

translated = point + translation

print(translated)
```

```python
def transformation_matrix(R, translation):

    T = np.eye(4)

    T[:3, :3] = R
    T[:3, :3] = translation

    return T
```

```python
point_h = np.array([
    10.0,
    0.0,
    0.0,
    1.0
])

R = rotation_z(90)

translation = np.array([
    100.0,
    200.0,
    300.0
])

T = transformation_matrix(R, translation)

result = T @ point_h

print(result)
```
